## 23. 建议练习顺序

> 来源：[Agent SDK overview](https://code.claude.com/docs/en/agent-sdk/overview)

1. 跑 §5「最小可运行示例」 `demo_query()`，只开 `Read`，先把消息流打印出来。
2. 跑 §7「读懂消息流」 `demo_read_stream()`，练**按块类型分流**——这是最该练熟的核心技能。
3. 把任务换成真实项目里的问题（"梳理认证流程""总结目录结构"）。
4. 跑 §15「多轮与打断」 `demo_multiturn()`，体会同上下文连续追问；再试 §13「会话」 的 resume / fork。
5. 跑 §10「自定义工具」 `demo_custom_tool()`，理解自定义工具怎么接进来、`is_error` 如何保住循环。
6. 加 §9「权限系统与 HITL」 的 `can_use_tool` 和 §11「Hooks」 的 Hook，把 agent 变"可控"。
7. 跑 §17 流式输出和 §18 结构化输出，这两个是接业务系统的关键形态。
8. 最后按 §21「生产化」 把预算、sandbox、可观测配齐，模拟一次生产部署。

> 只读文档不跑代码，不会真正建立对 Agent SDK 的手感。至少跑通一次 `query()` 和一次 `ClaudeSDKClient`。

**术语速查**

- **tool loop / agent loop**：模型请求调工具→执行→回传结果→再请求，循环到收尾。Agent SDK 的核心是把这层内建。
- **turn**：循环里一个完整的"请求 tool → 执行 → 回灌"周期；`max_turns` 只计带 tool call 的 turn。
- **进程内 MCP（SDK MCP server）**：把本地 Python 函数当工具暴露给 agent，同进程运行，无需外部服务。
- **streaming input mode**：`prompt` 传 async generator 的输入模式，附图/排队/打断/权限回调的前提。
- **HITL（Human-in-the-Loop）**：agent 在关键点停下等真人批准/输入，靠 `can_use_tool` 挂起 + `AskUserQuestion` 实现。
- **compaction**：context 逼近上限时 SDK 自动把旧历史压缩成摘要；持久规则要放 CLAUDE.md。
- **setting_sources**：文件系统配置（CLAUDE.md/skills/hooks/settings）的加载总开关；省略=全加载。

**Python 专属坑速查**（正文均已展开，集中列一遍）：SDK 不自动加载 `.env`（§3「安装与认证」）；`query()` 错误 result 后会 raise（§4「两个入口与两种输入模式」）；`can_use_tool` 需流式输入 + `{"continue_": True}` dummy hook（§9.3「can_use_tool 回调」）；被自动批准的调用到不了回调（§9.1「权限评估顺序」）；流式输入 generator 异常静默卡死（§16「流式输入」）；自定义工具抛异常整个循环终止（§10.5「错误处理铁律」）；进程内 server 不支持 `structuredContent`（§10.5「错误处理铁律」）；`AgentDefinition` 多词字段是 camelCase（§14.1「编程式定义」）；`SystemMessage` 的数据都在 `.data` dict 里（§7.1「Message 六种主类型」）；Python 无 `SessionStart`/`SessionEnd` 回调 hook（§11.1）、无 `auto` 权限模式（§9.2「permission_mode 全表」）、无 `persistSession: false`（§13.3「session_store」）。
